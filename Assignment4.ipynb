```python
import numpy as np
from scipy import stats

numbers = np.array([12, 15, 12, 18, 21, 12, 25, 30, 4, 9, 15])
print(f"Original Array: {numbers}\n")

mean_val = np.mean(numbers)
median_val = np.median(numbers)
mode_result = stats.mode(numbers, keepdims=True)
mode_val = mode_result.mode[0]
mode_count = mode_result.count[0]

max_val = np.max(numbers)
min_val = np.min(numbers)
sum_val = np.sum(numbers)
range_val = np.ptp(numbers)
variance_val = np.var(numbers)
std_dev_val = np.std(numbers)

q25 = np.percentile(numbers, 25)
q75 = np.percentile(numbers, 75)

geom_mean = stats.gmean(numbers)
harm_mean = stats.hmean(numbers)
mad = np.median(np.abs(numbers - np.median(numbers)))

print("--- Statistics ---")
print(f"Mean: {mean_val:.2f}")
print(f"Median: {median_val}")
print(f"Mode: {mode_val} (appears {mode_count} times)")
print(f"Maximum: {max_val}")
print(f"Minimum: {min_val}")
print(f"Sum: {sum_val}")
print(f"Range: {range_val}")
print(f"Variance: {variance_val:.2f}")
print(f"Standard Deviation: {std_dev_val:.2f}")
print(f"25th Percentile: {q25:.2f}")
print(f"75th Percentile: {q75:.2f}")
print(f"Geometric Mean: {geom_mean:.2f}")
print(f"Harmonic Mean: {harm_mean:.2f}")
print(f"Median Absolute Deviation: {mad:.2f}\n")

print("--- Arithmetic Operations (with constant = 5) ---")
print(f"Addition: {numbers + 5}")
print(f"Subtraction: {numbers - 5}")
print(f"Multiplication: {numbers * 5}")
print(f"Division: {np.round(numbers / 5, 2)}")
```
