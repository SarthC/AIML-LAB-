```python
import pandas as pd

data = {
    "Name": ["Sarthak", "Rahul", "Priya", "Sneha", "Akash"],
    "Marks": [85, 78, 92, 88, 95]
}

df = pd.DataFrame(data)

print(df)
print("\nAverage Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
```

```python
import matplotlib.pyplot as plt

names = ["Sarthak", "Rahul", "Priya", "Sneha", "Akash"]
marks = [85, 78, 92, 88, 95]

plt.bar(names, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

names = ["Sarthak", "Rahul", "Priya", "Sneha", "Akash"]
marks = [85, 78, 92, 88, 95]

sns.barplot(x=names, y=marks)

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks using Seaborn")
plt.show()
```
